**你的解答**：

```
1. softmax(e):
   分子: exp(2)≈7.39, exp(1)≈2.72, exp(0)=1
   分母: 7.39+2.72+1 = 11.11
   α = [0.665, 0.245, 0.090]

2. c = 0.665·[1,0] + 0.245·[0,1] + 0.090·[1,1]
     = [0.665+0.090, 0.245+0.090]
     = [0.755, 0.335]

3. 把第 3 个位置 score 设为 -inf：e = [2.0, 1.0, -inf]
   exp(-inf)=0，α = [exp(2)/(exp(2)+exp(1)),  exp(1)/(...),  0]
   ≈ [0.731, 0.269, 0]
```
